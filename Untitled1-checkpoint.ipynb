```python
import pandas as pd

df = pd.read_csv("DelayedFlights.csv")
df.head()
```

```python
df.isnull().sum()
```

```python
cols = ['CarrierDelay', 'WeatherDelay', 'NASDelay', 'SecurityDelay', 'LateAircraftDelay']

df = df.dropna(subset=cols, how='all')
```

```python
df.head()
```

```python
df = df.drop('CancellationCode', axis=1)
```

```python
df = df.drop('Cancelled', axis=1)
```

```python
df.isnull().sum()
```

```python
df = df[df['TailNum'].notna()]
```

```python
df.isnull().sum()
```

```python
time_cols = ['DepTime', 'CRSDepTime', 'ArrTime', 'CRSArrTime']

for col in time_cols:
    df[col] = df[col].astype(int).astype(str).str.zfill(4)
    df[col] = df[col].str[:2] + ':' + df[col].str[2:]
```

```python
df.tail()
```

```python
df['Date'] = pd.to_datetime(df[['Year', 'Month', 'DayofMonth']].rename(columns={'DayofMonth': 'Day'}))
```

```python
df.drop(columns=['Year', 'Month', 'DayofMonth'], inplace=True)
```

```python
df.info()
```

```python
df.drop(columns=['Unnamed: 0'], inplace=True)
```

```python
df.drop(columns=['TailNum'], inplace=True)
```

```python
df.info()
```

```python
df.to_csv('file_cleaned.csv', index=False)
```

```python
days = {1: 'Monday', 2: 'Tuesday', 3: 'Wednesday', 
        4: 'Thursday', 5: 'Friday', 6: 'Saturday', 7: 'Sunday'}

df['DayName'] = df['DayOfWeek'].map(days)

df.to_csv('file_cleaned.csv', index=False)
```
